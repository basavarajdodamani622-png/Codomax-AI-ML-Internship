```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Module 03: Data Analysis & Visualization")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Libraries imported successfully!")
```

```python
# Create a sample student performance dataset

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Study_Hours": [2, 4, 3, 5, 1, 6, 4, 2, 7, 3],
    "Attendance": [65, 80, 75, 90, 55, 95, 85, 60, 98, 70],
    "Marks": [45, 70, 60, 85, 35, 92, 78, 40, 96, 58]
}

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
display(df)
```

```python
# Explore the dataset

print("First 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nDataset Shape (rows, columns):")
print(df.shape)

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Remove duplicate rows, if any
df = df.drop_duplicates()

# Fill missing numeric values with their column averages
for col in ["Study_Hours", "Attendance", "Marks"]:
    df[col] = df[col].fillna(df[col].mean())

print("\nDataset after cleaning:")
display(df)

print("Final dataset shape:", df.shape)
```

```python
# Analyze student performance

print("Average Study Hours:", round(df["Study_Hours"].mean(), 2))
print("Average Attendance:", round(df["Attendance"].mean(), 2))
print("Average Marks:", round(df["Marks"].mean(), 2))

print("\nStudent with Highest Marks:")
display(df.loc[df["Marks"].idxmax()])

print("\nStudent with Lowest Marks:")
display(df.loc[df["Marks"].idxmin()])

print("\nCorrelation between Study Hours and Marks:")
print(round(df["Study_Hours"].corr(df["Marks"]), 2))
```

```python
# Bar chart: Marks obtained by each student

plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Marks"])

plt.title("Student Marks Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
# Bar chart: Study hours by student

plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Study_Hours"])

plt.title("Student Study Hours Comparison")
plt.xlabel("Student")
plt.ylabel("Study Hours per Week")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```
